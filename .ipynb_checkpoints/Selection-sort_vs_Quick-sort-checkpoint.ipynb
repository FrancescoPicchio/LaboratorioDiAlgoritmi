<h1 style="font-size: 5rem; color: darkblue">Selection-sort vs Quick-sort</h1>

# Introduzione
_Selection-sort_ e _Quick-sort_ sono tra i principali __algoritmi di ordinamento__ utilizzati per ordinare una lista di elementi in ordine __crescente__ o __decrescente__.
Come vedremo, nonostante _Selection-sort_ sia __semplice__ da implementare e da capire, è un algoritmo poco usato per la sua __inefficenza__, viceversa _Quick-sort_ si pone generalmente come il __migliore__ fra gli algoritmi di ordinamento da poter usare.

<h2 style="font-size: 3rem; color: #3437eb;">Selection-sort</h2>

## Breve descrizione
_Selection-sort_ funziona confrontando ogni elemento _uno a uno_ e trovando ripetutamente l'elemento __minimo__ dalla parte non ordinata della lista, dunque scambiandolo con l'elemento all'__inizio della parte non ordinata__. 
Questo processo viene ripetuto fino a quando l'intera lista è __ordinata__.

## Codice

```python
def selectionSort(lista):
    for i in range(len(lista)):
        indice_min = i
        
        for j in range(i+1, len(lista)):
            if lista[j] < lista[indice_min]: #ordino in ordine crescente
                indice_min = j
        #eseguo quindi lo swap
        lista[i], lista[indice_min] = lista[indice_min], lista[i]
    return lista
```

### Esempio di utilizzo

```python
lista = [42, 13, 33, 55, 66, 31, 89, 2]
print('la lista ordinata è:', selectionSort(lista))
```

## Tempo di Esecuzione
Siccome nel _caso peggiore_ e nel caso _migliore_ dobbiamo comunque iterare su tutti gli __n elementi__ della lista e ripetere il processo __n volte__, _Selection sort_ ha una complessità temporale di **_O(n^2)_**, il che significa che le sue prestazioni diminuiscono molto __rapidamente__ all'aumentare della dimensione della lista

<h2 style="font-size: 3rem; color: #3437eb;">Quick-sort</h2>

## Breve descrizione
_Quick-sort_ funziona selezionando un elemento **_pivot_** dalla lista e partizionando la lista in due sottoliste, una contenente elementi __più piccoli__ del pivot e l'altra contenente elementi __più grandi__ del pivot. Il processo viene ripetuto in modo __ricorsivo__ sulle due _sottoliste_ fino a quando l'intera lista è ordinata.

## Codice

```python
def quickSort(lista, low, high):
    if low < high:
        div = partition(lista, low, high) 
        # richiamo quick-sort a sinistra del punto di divisione
        quickSort(lista, low, div - 1)
        #richiamo quick-sort a destra del punto di divisione
        quickSort(lista, div + 1, high)

def partition(lista, low, high):
    # decido di scegliere come pivot l'ultimo elemento della lista
    pivot = lista[high]
    
    i = low - 1
    for j in range(low, high):
        if lista[j] <= pivot:
            # incremento l'indice e faccio lo swap degli elementi
            i += 1
            lista[i], lista[j] = lista[j], lista[i]
            
    # infine scambio il pivot con l'elemento successivo all'indice nel punto di divisione
    lista[i + 1], lista[high] = lista[high], lista[i + 1]  
    return i + 1
```

Nel codice sopra, la funzione _**partition**_ si occupa di dividere l'array in due parti, una con elementi **minori o uguali** al _pivot_ e l'altra con elementi **maggiori** del _pivot_, ritornando infine l'indice del punto di divisione.

### Esempio di utilizzo

```python
lista = [41, 22, 1, 24, 52, 13, 73, 98, 89]
quickSort(lista, 0, len(lista) - 1)
print('la lista ordinata è: ', lista)
```

## Tempo di Esecuzione

La _complessità computazionale_ dell'algoritmo di _quicksort_ dipende dalla __scelta del _pivot___ e dal modo in cui viene effettuato il _partition_. 

Infatti se il pivot viene scelto in **modo casuale** o se viene scelto un **pivot mediano** (ovvero l'elemento di mezzo del mio insieme di elementi), allora il *costo computazionale* dell'algoritmo sarà generalmente **_O(n log n)_** . Questo perché, in questi casi, ad ogni passo il numero di elementi viene diviso approssimativamente a metà, e il numero di volte che l'insieme viene diviso a ogni passo dell'algoritmo è __log n__, con _n_ numero di elementi, mentre il partition richiede un tempo **proporzionale** a n (in quanto scorre tutti gli elementi), per cui il **tempo di esecuzione** è dato dalla complessità di partition moltiplicata per il numero di volte che l'array viene diviso in parti, ovvero O(n log n). 


Tuttavia, se il pivot viene scelto in **modo non casuale** (ad esempio, l'elemento **più grande** o **più piccolo** della lista di elementi), allora il costo computazionale può diventare **_O(n^2)_**, il che rende l'algoritmo meno efficiente.

Riassumendo:
* **Caso migliore e caso medio**: O(n log n)
* **Caso peggiore**: O(n^2)

# Testing

```python
import random 
from timeit import default_timer as timer
import matplotlib.pyplot as plt
import numpy as np
import sys
sys.setrecursionlimit(10 ** 9)
    
def listaCasuale(num_elementi):
    # genera una lista di numeri interi casuali
    lista = list()
    for i in range(num_elementi):
        lista.append(random.randint(0, num_elementi * 10))
    return lista

def listaOrdinata(num_elementi):
    lista = list()
    for i in range(num_elementi):
        lista.append(i)
    return lista

def listaOrdinataCasualmente(num_elementi):
    # genero la lista di elementi ordinati da 0 a n
    lista = np.arange(0, num_elementi)
    #mescolo gli elementi in modo casuale
    np.random.shuffle(lista)
    return lista


def test(alg_lista, num_elementi, iterazioni):
    dimensioni_lista = []
    confronto = []
    selection_sort_t = []
    quick_sort_t = []
    n = 0
    lista_max = alg_lista(num_elementi * iterazioni)
    
    for i in range(iterazioni):
        n = num_elementi * (i + 1)
        dimensioni_lista.append(n)
        print(n)
        lista = lista_max[:n]
        copia_lista = lista.copy() # mi serve per fare il test successivo
        # esegui selection sort e misura il tempo di esecuzione ad ogni iterazione
        start = timer()
        selectionSort(lista)
        stop = timer()
        selection_sort_t.append((stop - start))
        # esegui quicksort e misura il tempo di esecuzione ad ogni iterazione
        start = timer()
        quickSort(copia_lista, 0, n - 1)
        stop = timer()
        quick_sort_t.append((stop - start))

    # crea un grafico che mostra l'andamento temporale dei due algoritmi
    plt.plot(dimensioni_lista, selection_sort_t, label='Selection-sort')
    plt.plot(dimensioni_lista, quick_sort_t, label='Quick-sort')
    plt.xlabel('Numero di elementi')
    plt.ylabel('Tempo di esecuzione (secondi)')
    plt.title('Andamento temporale di Selection-sort e Quick-sort')
    plt.legend()
    plt.show()
    
    
def main():        
    num_elementi = 1700
    iterazioni = 10
    test(listaCasuale, num_elementi, iterazioni)
    print('Secondo test...')
    test(listaOrdinata, num_elementi, iterazioni)
    print('Terzo test...')
    test(listaOrdinataCasualmente, num_elementi, iterazioni)
    
main()
```

Si vanno ad eseguire 3 test in sequenza:
* Nel primo caso passo in input ai due algoritmi una lista di elementi presi in modo __random__, aumentando il _range_ dei possibili valori moltiplicando per 10 il _numero di elementi_ della lista
* Nel secondo caso invece passeremo come input una lista di elementi __già ordinata__ (in ordine crescente) per vedere come si comporta nel suo _caso peggiore_ il _quick-sort_.
* Nel l'ultimo caso invece prendiamo una lista con elementi in ordine crescente da 0 a *num_elementi*, e tramite la funzione ___shuffle___ della libreria _numpy_, diamo un __ordine casuale__ agli elementi della lista.

## Conclusioni e commenti


Guardando i grafici si conferma quindi l'ipotesi iniziale, ovvero che _selection-sort_ rimane un algoritmo __inefficiente__ in quasi tutti i casi, addirittura anche nel caso di lista già ordinata si hanno stessi risultati!<br>Viceversa _quick-sort_ anche a fronte di un gran numero di elementi, come __10000 elementi__, ha mediamente dei tempi di esecuzione molto __più efficenti__ (sempre quasi nulli!) rispetto al suo avversario _selection-sort_.


Il caso di lista già ordinata invece però fà notare un grosso svantaggio per _quick-sort_, tanto da avere tempi **peggiori** di _selection-sort_ !<br>
Anche se questo è un caso limite (che comunque sminuisce il compito di ordinare un insieme di elementi), ha dato comunque iniziativa ad una serie di miglioramenti per l'algoritmo di _quick-sort_, come ad esempio la **scelta iterativa del pivot**, tale che la scelta non ricada sempre sullo stesso elemento e per fare in modo via via di dividere il più possibile gli elementi a metà, rientrando nel _caso medio_ per il _tempo di esecuzione_!

Viene da chiedersi infine se oltre alla sua **semplicità**, esiste qualche altro vantaggio per l'uso di _selection-sort_, anche se non mostrato direttamente, analizzando come opera l'algortmo, si nota che ciascun elemento viene spostato **al più una volta**, dunque nei casi di elementi di dimensioni notevoli oppure difficili da spostare, trova la sua applicazione _selection-sort_ che in termini di _costi per spostamento_ trova il suo principale vantaggio!
